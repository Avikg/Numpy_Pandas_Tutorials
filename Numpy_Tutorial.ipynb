**Dimensionality**

```python
a1=np.arange(10)
a2=np.arange(12, dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)

print("a1", a1)
print("----------------------")
print("a2", a2)
print("----------------------")
print("a3", a3)
```

```python
#ndim
print("a1-dim-", a1.ndim)
print("----------------------")
print("a2-dim-", a2.ndim)
print("----------------------")
print("a3-dim-", a3.ndim)
```

```python
#shape
print("a1-shape-", a1.shape)
print("----------------------")
print("a2-shape-", a2.shape)
print("----------------------")
print("a3-shape-", a3.shape)
```

```python
#size
print("a1-size-", a1.size)
print("----------------------")
print("a2-size-", a2.size)
print("----------------------")
print("a3-size-", a3.size)
```

```python
#itemsize    one nt/float how much memory allocated 'per item' (In colab int/float take same size i.e 1 Byte)
print("a1-itemsize-", a1.itemsize)
print("----------------------")
print("a2-itemsize-", a2.itemsize)
print("----------------------")
print("a3-itemsize-", a3.itemsize)


a1=np.arange(10, dtype=np.int32)
print("------------------------After Int 32-----------------------")
print("a1-itemsize-", a1.itemsize)
print("----------------------")
print("a2-itemsize-", a2.itemsize)
print("----------------------")
print("a3-itemsize-", a3.itemsize)
```

```python
#dtype
print("a1-dtype-", a1.dtype)
print("----------------------")
print("a2-dtype-", a2.dtype)
print("----------------------")
print("a3-dtype-", a3.dtype)
```

# **Changing Datatype**

**Array Operations**

```python
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)

print('a1-', a1)
print('-----------------')
print('a2-', a2)
```

Scalar Operations

```python
#arithmetic
print(a1)
print('-----------------')



print('a1*2-', a1*2)
print('-----------------')
print('a1+2-', a1+2)
print('-----------------')
print('a1-2-', a1-2)
print('-----------------')
print('a1/2-', a1/2)
print('-----------------')
print('a1%2-', a1%2)
```

```python
#relational
print(a2)
print('-----------------')


print('a2>5-', a2>5)
print('-----------------')
print('a2>25-', a2>25)
print('-----------------')
print('a2>15-', a2>15)
print('-----------------')
print('a2==15-', a2==15)
print('-----------------')
print('a2!=15-', a2!=15)
```

Vector Operations

```python
#arithmetic

print('a1+a2-\n', a1+a2)
print('-----------------')
print('a1*a2-\n', a1*a2)
```

```python
#dot product    a.b^T
a=np.array([1,2,3])
b=np.array([4,5,6])
print(a.shape)
print(b.shape)
print(np.dot(a,b))
print(a@b)   ## Can be used for Matrix Multiplication also
```
