**Indexing and Slicing**

```python
a1=np.arange(10)
print('a1\n', a1)
a2=np.arange(12).reshape(3,4)
print('a2\n', a2)
a3=np.arange(8).reshape(2,2,2)
print('a3\n', a3)
```

```python
#lastitem/firstitem of a1
print('last item of a1-', a1[-1])
print('first item of a1-', a1[0])


#2d array
#I want 6
print(a2[1, 2])

##I want 4
print(a2[1, 0])



#3d array
#I want 5
print(a3[1, 0, 1])

#I want 2
print(a3[0, 1, 0])

#I want 0
print(a3[0, 0, 0])
```

```python
#slicing
print(a1)

#2,3,4
print(a1[2:5])
print(a1[2:5:2])

print('-----------')

#for 2d
print(a2)
#for first row
print(a2[0, :])
#for third column
print(a2[:, 2])
# for 5,6, 9,10
print(a2[1:, 1:3])
# for 0,3,8,11
print(a2[::2, ::3])
# for 1,3,9,11
print(a2[::2, 1::2])
# for 4,7
print(a2[1, ::3])

print('-----------')

#for 3d
a3=np.arange(27).reshape(3,3,3)   #contains 3 nos of 2-d arrays of size(3,3)
print(a3)
print('-------Operations------')
#for middle 2d array
print(a3[1, :, :])
#for middle 1st and last array
print(a3[::2, :, :])  #or print(a3[::2])
#1st one 2d array ka 2nd row
print(a3[0, 1]) #or print(a3[0, 1, :])
#2nd one 2d array ka middle column
print(a3[1, ::, 1]) #or print(a3[1,:, 1])
#for 22, 23, 25,26
print(a3[2, 1:, 1:])
#for 0,2,18,20
print(a3[::2, 0, ::2])
```

**Iterating**

```python
print('a1\n', a1)
print('a2\n', a2)
print('a3\n', a3)
```

```python
#for 1d
for i in a1:
  print(i, end=" ")

print('\n------------')

#for 2d
for i in a2:
  print(i)

print('------------')

#for 3d
for i in a3:
  print(i)

print('------------')

#for each item for 1d, 2d and 3d
for i in np.nditer(a3):
  print(i, end=" ")
```

**Stacking**

```python
#horizontal stacking---------2X2 and 2X2 two arrays(any no arrays possible), after op, 2X4
a4=np.arange(12).reshape(3,4)
a5=np.arange(12, 24).reshape(3,4)
print(a4)
print(a5)
print('------Afetr Operation------')
print(np.hstack((a4,a5,a4,a5)))
```

```python
#vertical stacking----------2X2 and 2X2 two arrays(any no arrays possible), after op, 4X2
print(np.vstack((a4,a5,a4,a5)))
```

**Splitting**
splitting opposite of stacking

```python
#horizontal splitting
print(a4)
print(np.hsplit(a4, 2))  # cut into 2 parts
print(np.hsplit(a4, 4))  # cut into 4 parts

for i in np.hsplit(a4, 4):
  print(i)
```

```python
#vertical splitting
print(a5)
print(np.vsplit(a5, 3))  # cut into 3 parts
#print(np.vsplit(a5, 4))  # cut into 4 parts
```
